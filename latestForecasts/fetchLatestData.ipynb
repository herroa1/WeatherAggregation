# Fetch latest data from Met Office Weather DataHub

Uses the Atmospheric Models API (see [weather_datahub_utilities](https://github.com/MetOffice/weather_datahub_utilities)) via `MetOfficeLoader` in [`loaders/met_office.py`](../loaders/met_office.py). The API key is read from the project-root `.env` (`MET_OFFICE_API_KEY`) — don't print it, as outputs are saved in the notebook. Files are saved under `data/` next to this notebook; downloads count towards the data allowance, so files already on disk are reused.

The API only serves the latest run(s) of each order; for historical data see `../archiveForecasts/testMetOffice.ipynb`.

```python
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path

import matplotlib.pyplot as plt

# Make the project-root loaders package importable from this folder
sys.path.insert(0, str(Path.cwd().parent))
from loaders import MetOfficeLoader

loader = MetOfficeLoader(data_dir="data")
```

```python
# List active orders on the account
orders = loader.list_orders()
orders
```

```python
# List files in the latest run of an order
ORDER_ID = orders[0]["orderId"]

files = loader.list_latest_files(ORDER_ID)
[f["fileId"] for f in files]
```

```python
# Download (if needed) and open each file in the latest run
uk = loader.load_latest(ORDER_ID, files)
list(uk)
```

```python
uk["ground_rainfall-accumulation-1h_+00"]
```

```python
# Rainfall over the first 6 forecast hours
rain = uk["ground_rainfall-accumulation-1h_+00"]["tirf"]
rain.isel(step=slice(0, 6)).plot(x="longitude", y="latitude", col="step", col_wrap=3, cmap="Blues", robust=True)
plt.show()
```
